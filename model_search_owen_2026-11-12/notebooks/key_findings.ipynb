## Key findings

**What we predict and how we score it**

1. <span style="color:#1f5fbf">[K1] We predict P(at least one bike at t + h) for each bike type, because whether a bike is there is what a rider needs to know.</span>
2. <span style="color:#1f5fbf">[K2] The score is the skill against a reference forecast, skill = 1 − Brier_model / Brier_ref, with calibrated persistence (an honest 'what you see now' probability) as the reference: 0 = no better than the reference, 1 = perfect.</span>

**What the model learns, and where it helps (validation days)**

3. <span style="color:#1f5fbf">[K3] Knowing the station matters most: the chosen model, which knows it, reaches 0.154 / 0.117 (classic / e-bike); LightGBM without any station information reaches 0.090 / 0.071.</span>
4. <span style="color:#1f5fbf">[K4] The model helps most on the last bike: skill 0.182 / 0.159 with one bike now, against 0.038 / 0.039 with 6 or more.</span>
5. <span style="color:#1f5fbf">[K5] The gain over the reference grows with the horizon, from 0.079 / 0.058 at h = 2 min to 0.177 / 0.143 at h = 14 min. The forecast itself gets less accurate further ahead (Brier 0.0103 to 0.0283, classic), but the reference degrades faster.</span>

**How well it carries over, and what comes next**

6. <span style="color:#1f5fbf">[K6] Test is lower than validation: 0.125 / 0.092 against 0.154 / 0.117. The test days include a weekend and the validation days do not; leaving the weekend out closes 29 % / 39 % of the gap, so the weekend is about a third of the drop.</span>
7. <span style="color:#1f5fbf">[K7] More training data helps, with a slowing gain (earlier count model, MSE skill, data-v1): 0.072 with 5 training days, 0.083 with 10, 0.109 with 20 (classic); about +0.019 per extra week over 10–20 days and +0.009 over 15–20. Data v2 (more days) tests this on the probability model.</span>
